```python
# -*-coding: utf-8 -*-
# 先运行 word_seg进行中文分词，然后再进行word_similarity计算
# 将Word转换成Vec，然后计算相似度 
from gensim.models import word2vec
import multiprocessing
import os
```

```python

# 如果目录中有多个文件，可以使用PathLineSentences
segment_folder = './journey_to_the_west/segment'
# 切分之后的句子合集
sentences = word2vec.PathLineSentences(segment_folder)
```

```python
sentences
```

```python

# 设置模型参数，进行训练
model = word2vec.Word2Vec(sentences, vector_size=100, window=3, min_count=1)
model
```

```python
print(model.wv['孙悟空'])
print(model.wv.similarity('孙悟空', '猪八戒'))
print(model.wv.similarity('孙悟空', '孙行者'))
print(model.wv.most_similar(positive=['孙悟空', '唐僧'], negative=['孙行者']))
```

* 参数解释

|                 参数                  |        含义        |                        本代码设置说明                        |
| :-----------------------------------: | :----------------: | :----------------------------------------------------------: |
|              `sentences`              |     训练数据集     | 由`PathLineSentences`读取文件夹下所有分词后的文本，每行词语空格分隔，是模型的训练原料 |
|           `vector_size=128`           |     词向量维度     | 把每一个词语映射成一串长度为 128 的浮点数数组；维度越高承载的语义信息越多，训练耗时、内存占用也更大（常用：100/128/200） |
|              `window=5`               | 上下文滑动窗口大小 | 训练时，某个词只会选取**左右各 5 个词语**作为上下文环境来学习词语之间的关系。窗口越大，能捕获远距离的语义关联，窗口小更侧重就近搭配习惯 |
|             `min_count=5`             |  词语最低出现频次  | 整部西游记里**出现次数少于 5 次的字词直接丢弃，不参与训练**。过滤掉生僻字、错别字、偶然出现的零碎字符，减少训练噪声，让模型向量更加稳定可靠 |
| `workers=multiprocessing.cpu_count()` |    训练的线程数    | `multiprocessing.cpu_count()`自动读取电脑 CPU 核心数量，利用多核 CPU 并行运算，大幅缩短模型训练耗时 |





model2.wv：模型的词向量存储空间，所有词语的向量都存在 wv 中
similarity(词1,词2)：计算两个词向量的余弦相似度
输出范围：[-1 ~ 1]，数值越靠近 1，代表两个词语在原文里语义、语境越相近。孙悟空和猪八戒同为徒弟，相似度数值一般偏高

向量数学运算公式：孙悟空向量 + 唐僧向量 − 孙行者向量
positive：传入的词语做向量相加
negative：传入的词语做向量相减
most_similar()：在全部词向量里，找出和运算结果向量最接近的一批词汇并返回

```python
# 设置模型参数，进行训练
model2 = word2vec.Word2Vec(sentences, vector_size=128, window=5, min_count=5, workers=multiprocessing.cpu_count())
# 保存模型
# ==========新增这两行==========
save_dir = "./models"
os.makedirs(save_dir, exist_ok=True)  # exist_ok=True：文件夹存在也不会报错
model2.save('./models/word2Vec.model')
print(model2.wv.similarity('孙悟空', '猪八戒'))
print(model2.wv.similarity('孙悟空', '孙行者'))
print(model2.wv.most_similar(positive=['孙悟空', '唐僧'], negative=['孙行者']))
```
